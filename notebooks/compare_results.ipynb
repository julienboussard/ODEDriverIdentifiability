```python
files_path = Path("../data/simple/noise=0.00_confounder=True/eval/ode_sparse_discovery")  # (you can change this to your own path)
files = glob.glob(str(files_path / "*.npz"))
```

```python
auprc_mean = 0
shd_mean = 0
auroc_mean = 0

n_files = len(files)

for file in files: 
    # Because HP search was done on lorenz84 which is the first here...
    print(file)
    result = np.load(file)
    print("AUROC ", result["joint_auroc"])
    if np.isnan(result["joint_auroc"]):
        auroc_mean += 1/n_files
    else:
        auroc_mean += result["joint_auroc"]/n_files
    print("AUPRC ", result["joint_auprc"])
    auprc_mean += result["joint_auprc"]/n_files
    print("SHD ", result["joint_shd"])
    shd_mean += result["joint_shd"]/n_files
    
print(f"AUROC {auroc_mean}, AUPRC {auprc_mean}, SHD {shd_mean}")
    
    
```
